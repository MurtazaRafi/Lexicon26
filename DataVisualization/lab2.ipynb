```python
import pandas as pd

listings = pd.read_csv('data/listings.csv') # load the file

listings['mask'] = listings['number_of_reviews'] == 0

print(listings['mask'].sum(), listings['mask'].mean())






```

By each neighbourhood

```python


listings.groupby('neighbourhood_cleansed').agg(n=('id', 'count'), share=('mask', 'mean')).sort_values('share', ascending=False)
```

## 2.


```python
listings['reviewed'] = listings['number_of_reviews'] > 0

divided = pd.qcut(listings['number_of_reviews'], 4, labels=['few', 'some', 'many', 'most'])

divided.value_counts()

pd.qcut(listings['number_of_reviews'], 4).cat.categories # the borders of the groups

listings['number_of_reviews'].describe()


reviewed = listings[listings['number_of_reviews'] >= 1].copy()
reviewed['review_group'] = pd.qcut(
    reviewed['number_of_reviews'], 4,
    labels=['few', 'some', 'many', 'most']
)

reviewed['review_group']
```
