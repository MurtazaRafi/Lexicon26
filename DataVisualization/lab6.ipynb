```python
listings['number_of_reviews'].describe()
```

```python
listings['number_of_reviews'].quantile([0.1, 0.25, 0.5, 0.75, 0.9, 0.99]).round(0)
```

```python
q1 = listings['number_of_reviews'].quantile(0.25)
q3 = listings['number_of_reviews'].quantile(0.75)

iqr = q3 - q1

print(round(q1, 1), round(q3, 1), round(iqr, 1))
```

What is the difference between the 25th to 75th percentage of the spread of the values

10 %

```python
# Calculate the two fences using IQR rule

upper = q3 + 1.5 * iqr
lower = q1 - 1.5 * iqr

print(round(lower, 1), round(upper, 1))
```

```python
outliers = listings[listings['number_of_reviews'] > upper]

print(len(outliers))
outliers['number_of_reviews'].describe().round(0)
```

```python
print(outliers['room_type'].value_counts())
print(outliers['accommodates'].median(), listings['accommodates'].median())
print((outliers['accommodates'] >= 6).mean(), (listings[listings['price'] <= upper]['accommodates'] >= 6).mean())
```
