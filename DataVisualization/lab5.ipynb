```python
import pandas as pd

rentals_raw = pd.read_csv('data/bike_rentals.csv')

rentals = rentals_raw.copy()

print(rentals.shape)
print(rentals.dtypes)
```

```python
empty = rentals.columns[rentals.isna().all()] # names of the empty columns

empty
```

```python
rentals = rentals.dropna(axis=1, how='all') # remove the empty columns

print(rentals.shape)
print(rentals.isna().all().sum())
```

2.

```python
rentals['price'].head(3)
```

```python
price = rentals['price'].str.replace(',', '')
price.head(3)
```

```python
price = price.str.replace('SEK', '', regex=False)
price.head(3)
```

```python
rentals['price'] = price.astype(float)

rentals['price'].dtype
```

```python
rentals['price'].describe().round(1)
```

```python

rentals.sort_values('price', ascending=False).head(3)[['rental_id', 'bike_type', 'status' ,'price']]

# rental_id, bike_type, status price
```
